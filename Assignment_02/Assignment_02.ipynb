```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.cbook import get_sample_data
from PIL import Image
```

```python
image_path = get_sample_data("grace_hopper.jpg", asfileobj=False)
with Image.open(image_path) as source:
    image = np.asarray(source.convert("RGB"))
```

```python
# Question 1
mean_gray = np.rint(image.mean(axis=2)).astype(np.uint8)

weights = np.array([0.7, 0.2, 0.1])
weighted_gray = np.rint(image @ weights).astype(np.uint8)

fig, axes = plt.subplots(1, 3, figsize=(12, 4), layout="constrained")
axes[0].imshow(image)
axes[0].set_title("Original RGB")
for ax, result, title in zip(
    axes[1:], (mean_gray, weighted_gray), ("Channel mean", "Weighted grayscale")
):
    ax.imshow(result, cmap="gray", vmin=0, vmax=255)
    ax.set_title(title)
for ax in axes:
    ax.axis("off")
plt.show()
```
