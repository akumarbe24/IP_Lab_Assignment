```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.cbook import get_sample_data
from PIL import Image
```

```python
image_path = get_sample_data("grace_hopper.jpg", asfileobj=False)
with Image.open(image_path) as source:
    image = np.asarray(source.convert("L"))
```

```python
# Question 1
mean_threshold = image.mean()
mean_binary = np.where(image > mean_threshold, 255, 0).astype(np.uint8)

fixed_threshold = 127
fixed_binary = np.where(image > fixed_threshold, 255, 0).astype(np.uint8)

fig, axes = plt.subplots(1, 3, figsize=(12, 4), layout="constrained")
for ax, result, title in zip(
    axes,
    (image, mean_binary, fixed_binary),
    ("Grayscale", f"Mean threshold ({mean_threshold:.2f})", f"Fixed threshold ({fixed_threshold:g})"),
):
    ax.imshow(result, cmap="gray", vmin=0, vmax=255)
    ax.set_title(title)
    ax.axis("off")
plt.show()
```
