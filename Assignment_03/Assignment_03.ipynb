```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.cbook import get_sample_data
from PIL import Image
```

```python
image_path = get_sample_data("grace_hopper.jpg", asfileobj=False)
with Image.open(image_path) as source:
    image = np.asarray(source.convert("L"))

binary = np.where(image > 127, 255, 0).astype(np.uint8)
```

```python
# Question 1
width = 5
gray_color = 128
binary_color = 255

padded_gray = np.pad(image, width, mode="constant", constant_values=gray_color)
padded_binary = np.pad(binary, width, mode="constant", constant_values=binary_color)

fig, axes = plt.subplots(2, 2, figsize=(8, 8), layout="constrained")
for ax, result, title in zip(
    axes.flat,
    (image, padded_gray, binary, padded_binary),
    ("Grayscale", "Padded grayscale", "Binary", "Padded binary"),
):
    ax.imshow(result, cmap="gray", vmin=0, vmax=255)
    ax.set_title(title)
    ax.axis("off")
plt.show()
```

```python
# Question 2
gray_complement = 255 - image
binary_complement = 255 - binary

fig, axes = plt.subplots(2, 2, figsize=(8, 8), layout="constrained")
for ax, result, title in zip(
    axes.flat,
    (image, gray_complement, binary, binary_complement),
    ("Grayscale", "Grayscale complement", "Binary", "Binary complement"),
):
    ax.imshow(result, cmap="gray", vmin=0, vmax=255)
    ax.set_title(title)
    ax.axis("off")
plt.show()
```
